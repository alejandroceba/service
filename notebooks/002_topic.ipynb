```python
import matplotlib.pyplot as plt

# Define a color palette for all divisions
colors = plt.cm.tab20(range(len(divisiones)))

fig, ax = plt.subplots()

for i, division in enumerate(divisiones):
    mask = df1.division == division
    ax.scatter(df1.loc[mask, "x_long"], df1.loc[mask, "y_lat"], 
               s=1, color=colors[i], label=division)

ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.legend(bbox_to_anchor=(1.05, 1),loc='best', markerscale=5)
plt.show()
```
