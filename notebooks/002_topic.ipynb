```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
f = "../data/data_sen_cegd_MW_2017_2024.csv"
df1 = pd.read_csv(f)
df1
```

```python
fig, ax = plt.subplots()

ax.scatter(df1.x_long,df1.y_lat,s=1)

plt.show()
```

```python
divisiones = df1.division.unique()
len(divisiones)
```

```python
import matplotlib.pyplot as plt

# Define a color palette for all divisions
colors = plt.cm.tab20(range(len(divisiones)))

fig, ax = plt.subplots()

for i, division in enumerate(divisiones):
    mask = df1.division == division
    ax.scatter(df1.loc[mask, "x_long"], df1.loc[mask, "y_lat"], 
               s=1, color=colors[i], label=division)

ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.legend(bbox_to_anchor=(1.05, 1),loc='best', markerscale=5)
plt.show()
```

```python
years = np.arange(2017, 2025)

fig, ax = plt.subplots()
for year in years:
    x = np.full(len(df1), year)
    ax.scatter(x, df1[str(year)], s=1)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

for idx, row in df1.iterrows():
    ax.plot(years, df1.loc[idx, years.astype(str)], linewidth=0.5, alpha=0.7)

ax.set_xlabel('Year')
ax.set_ylabel('MW')
ax.set_title('Node Values Over Time (2017-2024)')
plt.show()
```
